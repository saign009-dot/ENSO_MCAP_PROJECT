# Makes a table of the data from ghcnd_station_download_v1.0 and ENSO_indicies_download_v1.0

This script take daily observation data from the Station Data/ghcnd  
and index data from Indicies Data/El_Nino_Indicies_firstpass  
Then makes a built-out version of this table -> https://www.dnr.state.mn.us/climate/journal/el-nino-2026.html  

Note that averaging stations gives a station average not necesarilly a Minnesota average  
    more stations will make the avarage more representative of the state  

v0.0 designates this is a work in progress  
v1.0 should contribute to a reproducible workflow on any machine with the other 1.0 notebooks
